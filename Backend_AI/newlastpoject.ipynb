```python
def preprocess_data(df_to_process: pd.DataFrame) -> pd.DataFrame:
    df_to_process = df_to_process.copy()

    if "sound_effects" in df_to_process.columns and "sound_effects_list" not in df_to_process.columns:
        df_to_process["sound_effects_list"] = (
            df_to_process["sound_effects"]
            .fillna("")
            .astype(str)
            .apply(lambda x: [item.strip() for item in x.split(",") if item.strip()])
        )

    if "chapter" in df_to_process.columns and "chapter_num" not in df_to_process.columns:
        df_to_process["chapter_num"] = df_to_process["chapter"].astype(str).str.extract(r"(\d+)\.")

    if "chapter" in df_to_process.columns and "chapter_title" not in df_to_process.columns:
        df_to_process["chapter_title"] = (
            df_to_process["chapter"]
            .astype(str)
            .apply(lambda x: x.split(".", 1)[1].strip() if "." in x else x.strip())
        )

    return df_to_process


def load_and_preprocess_dataset(grade_level: str) -> pd.DataFrame:
    global df
    file_path = GRADE10_CSV if grade_level == "Grade 10" else GRADE11_CSV
    df = pd.read_csv(file_path, encoding="latin-1")
    df = preprocess_data(df)
    return df
```

```python
class SoundEffectsMixer:
    def __init__(self, sounds_folder: str):
        self.sounds_folder = sounds_folder
        self.sound_effects = {}
        self.load_sound_effects()

    def load_sound_effects(self):
        self.sound_effects = {}
        if not os.path.exists(self.sounds_folder):
            print(f"Warning: sounds folder not found: {self.sounds_folder}")
            return

        for sound_file in os.listdir(self.sounds_folder):
            if sound_file.lower().endswith((".mp3", ".wav", ".ogg")):
                name = os.path.splitext(sound_file)[0]
                path = os.path.join(self.sounds_folder, sound_file)
                try:
                    self.sound_effects[name] = AudioSegment.from_file(path)
                except Exception as e:
                    print(f"Failed to load sound effect '{sound_file}': {e}")
```

```python
import requests
import base64
import io
import soundfile as sf # Explicitly import soundfile for API response handling

class ApiTTSWrapper:
    def __init__(self, api_url):
        self.api_url = api_url
        self._sample_rate = 22050 # Default sample rate, can be updated by API response

    def tts(self, text, length_scale=1.0, lang="en-IN"): # lang="en-IN" for Indian accent
        payload = {"text": text, "lang": lang}

        try:
            r = requests.post(self.api_url, json=payload, timeout=60)
            r.raise_for_status() # Raise an exception for HTTP errors (4xx or 5xx)
            data = r.json()
            audio_base64 = data.get("audio_base64")
            if not audio_base64:
                raise ValueError("API response did not contain 'audio_base64'.")

            audio_bytes = base64.b64decode(audio_base64)
            audio_stream = io.BytesIO(audio_bytes)
            waveform, sr = sf.read(audio_stream)

            if sr != self._sample_rate:
                # Update sample rate if the API returns a different one
                self._sample_rate = sr

            return waveform.astype(np.float32)
        except requests.exceptions.ConnectionError:
            print(f"Error: Could not connect to API at {self.api_url}. Is the backend running and reachable?")
            raise
        except requests.exceptions.Timeout:
            print(f"Error: API request timed out for {self.api_url}.")
            raise
        except requests.exceptions.RequestException as e:
            print(f"API call failed: {e}")
            raise
        except Exception as e:
            print(f"Error processing API response: {e}")
            raise

    @property
    def synthesizer(self):
        # Mimic the Coqui TTS synthesizer structure to maintain compatibility
        class DummySynthesizer:
            @property
            def tts_model(self):
                class DummyModel:
                    @property
                    def config(self):
                        class DummyConfig:
                            @property
                            def audio(
                                self,
                            ):
                                class DummyAudio:
                                    sample_rate = self._sample_rate
                                return DummyAudio()
                        return DummyConfig()
                return DummyModel()

        return DummySynthesizer()


def init_tts_model():
    global tts_model

    if "tts_model" in globals() and tts_model is not None:
        # If an ApiTTSWrapper is already initialized with a valid API_URL, return it
        if isinstance(tts_model, ApiTTSWrapper) and API_URL and API_URL != "https://your-backend-url/api/tts":
            print("API TTS model already initialized.")
            return tts_model

    if API_URL and API_URL != "https://your-backend-url/api/tts":
        print(f"Attempting to use external VIS_Backend at: {API_URL}")
        try:
            tts_model = ApiTTSWrapper(API_URL)
            print("Successfully configured for external VIS_Backend.")
            return tts_model
        except Exception as e:
            print(f"Failed to initialize external VIS_Backend: {e}")
            print("Falling back to dummy audio generator.")
            # Fall through to dummy model if API fails

    print("API_URL is not configured or external API failed. Using fallback dummy audio generator.")

    class DummyTTS:
        def tts(self, text, **kwargs):
            return np.zeros(22050 * 2, dtype=np.float32)

        @property
        def synthesizer(self):
            class DummySynthesizer:
                @property
                def tts_model(self):
                    class DummyModel:
                        @property
                        def config(self):
                            class DummyConfig:
                                @property
                                def audio(
                                    self,
                                ):
                                    class DummyAudio:
                                        sample_rate = 22050
                                    return DummyAudio()
                            return DummyConfig()
                    return DummyModel()

            return DummySynthesizer()

    tts_model = DummyTTS()

    return tts_model
```

```python
grade_selector = widgets.RadioButtons(
    options=["Grade 10", "Grade 11"],
    value="Grade 10",
    description="Grade:",
    disabled=False,
    layout=widgets.Layout(width="auto"),
)

chapter_dropdown = widgets.Dropdown(
    options=[],
    description="Chapter:",
    disabled=False,
)

lesson_dropdown = widgets.Dropdown(
    options=[],
    description="Lesson:",
    disabled=False,
)

emotion_slider = widgets.FloatSlider(
    value=1.0,
    min=0.5,
    max=2.0,
    step=0.1,
    description="Emotion Intensity:",
    disabled=False,
    continuous_update=False,
    orientation="horizontal",
    readout=True,
    readout_format=".1f",
)

sound_toggle = widgets.ToggleButtons(
    options=["With Effects", "Only Effects", "Without Effects"],
    description="Sound:",
    disabled=False,
    button_style="",
    tooltips=["Play with sound effects", "Play only sound effects", "Play narration only"],
)

play_button = widgets.Button(
    description="🎵 Play Lesson",
    disabled=False,
    button_style="success",
    tooltip="Play the selected lesson",
    icon="play",
)
```

```python
def update_lessons(change):
    chapter_value = change["new"] if isinstance(change, dict) else chapter_dropdown.value
    if not chapter_value:
        lesson_dropdown.options = []
        return

    chapter_num = str(chapter_value).split(".")[0].strip()

    if "df" in globals() and isinstance(df, pd.DataFrame) and not df.empty:
        lessons = (
            df[df["chapter_num"].astype(str) == chapter_num]["Grade/Topic"]
            .dropna()
            .astype(str)
            .tolist()
        )
        lesson_dropdown.options = lessons
        if lessons:
            lesson_dropdown.value = lessons[0]
    else:
        lesson_dropdown.options = []


def on_grade_change(change):
    global df
    selected_grade = change["new"] if isinstance(change, dict) else grade_selector.value
    print(f"Loading data for {selected_grade}...")

    df = load_and_preprocess_dataset(selected_grade)

    chapter_df = df[["chapter_num", "chapter_title"]].dropna().drop_duplicates()
    chapter_options = [f"{row['chapter_num']}. {row['chapter_title']}" for _, row in chapter_df.iterrows()]
    chapter_dropdown.options = chapter_options

    if chapter_options:
        chapter_dropdown.value = chapter_options[0]
        update_lessons({"new": chapter_options[0]})
    else:
        lesson_dropdown.options = []
```

```python
import re

def get_sample_rate() -> int:
    if isinstance(tts_model, TTS):
        return tts_model.synthesizer.tts_model.config.audio.sample_rate
    return 22050


def generate_tts_wav(text: str, length_scale: float = 1.0) -> str:
    sample_rate = get_sample_rate()
    waveform_np = tts_model.tts(text=text, length_scale=length_scale)

    if isinstance(waveform_np, list):
        waveform_np = np.array(waveform_np, dtype=np.float32)

    waveform_np = np.asarray(waveform_np, dtype=np.float32)
    waveform_tensor = torch.from_numpy(waveform_np).unsqueeze(0)

    with tempfile.NamedTemporaryFile(delete=False, suffix=".wav") as f:
        temp_path = f.name
        torchaudio.save(temp_path, waveform_tensor, sample_rate)
    return temp_path


def parse_sound_annotations(annotation_text: str) -> dict:
    keyword_to_sound_map = {}
    if pd.isna(annotation_text) or not str(annotation_text).strip():
        return keyword_to_sound_map

    # Split by semicolon or newline characters to correctly parse annotations
    for entry in re.split(r'[;\n]', str(annotation_text)):
        entry = entry.strip()
        if ":" not in entry:
            continue
        sound_name, keywords_str = entry.split(":", 1)
        keywords = [k.strip().lower() for k in keywords_str.split("|") if k.strip()]
        for keyword in keywords:
            keyword_to_sound_map[keyword] = sound_name.strip()

    return keyword_to_sound_map


def build_lesson_segments(text_to_speak: str, keyword_to_sound_map: dict):
    sentences = nltk.sent_tokenize(text_to_speak)
    lesson_segments = []

    for sentence in sentences:
        associated_sound_effects = []
        for keyword, sound_name in keyword_to_sound_map.items():
            if keyword in sentence.lower():
                associated_sound_effects.append(sound_name)
        associated_sound_effects = list(dict.fromkeys(associated_sound_effects))
        lesson_segments.append(
            {
                "sentence": sentence,
                "associated_sound_effects": associated_sound_effects,
            }
        )

    return lesson_segments
```

```python
def play_selected_lesson(_):
    clear_output(wait=True)
    display(widgets.VBox([grade_selector, chapter_dropdown, lesson_dropdown, emotion_slider, sound_toggle, play_button]))

    if not chapter_dropdown.value or not lesson_dropdown.value:
        print("Please select a chapter and lesson.")
        return

    chapter_num = str(chapter_dropdown.value).split(".")[0].strip()
    lesson_topic = lesson_dropdown.value

    matched_rows = df[
        (df["chapter_num"].astype(str) == chapter_num)
        & (df["Grade/Topic"].astype(str) == str(lesson_topic))
    ]

    if matched_rows.empty:
        print("No lesson found for the selected chapter/topic.")
        return

    lesson_data = matched_rows.iloc[0]
    sound_mode = sound_toggle.value
    sample_rate = get_sample_rate()

    keyword_to_sound_map = parse_sound_annotations(lesson_data.get("sound_annotations", ""))
    print(f"\nParsed keyword to sound map: {keyword_to_sound_map}")
    print(f"\n📚 Chapter: {lesson_data['chapter']}")
    print(f"🎯 Topic: {lesson_data['Grade/Topic']}")
    print(f"😊 Emotion: {lesson_data.get('emotion', 'N/A')}")
    print(f"🔊 Sound Effects: {lesson_data.get('sound_effects', '')}")

    text_to_speak = str(lesson_data.get("simplified_text", "")).replace("e.g.", "example is")
    lesson_segments = build_lesson_segments(text_to_speak, keyword_to_sound_map)

    print("\n--- Lesson Segments with Associated Sound Effects ---")
    for segment in lesson_segments:
        print(f"Sentence: {segment['sentence']}")
        print(f"Sounds: {segment['associated_sound_effects']}")
        print("-" * 20)
    print("--------------------------------------------------")

    # Emotion slider kept as requested. Here it controls speaking speed mildly.
    # Lower value = slower, higher value = slightly faster.
    length_scale = max(0.8, min(1.4, 1.2 - ((emotion_slider.value - 1.0) * 0.2)))

    chapter_title_clean = str(lesson_data.get("chapter_title", "")).strip()
    topic_clean = str(lesson_data.get("Grade/Topic", "")).split(":")[-1].strip()

    intro_part1_text = f"Hi...! you Selected Chapter is {chapter_title_clean} and you choose topic is {topic_clean}. "
    intro_part2_text = f"Today, I am going to discuss about {topic_clean}. "

    current_effects_list = list(lesson_data.get("sound_effects_list", []))

    print("\n🔊 Generating introductory speech part 1...")
    intro_part1_path = generate_tts_wav(intro_part1_text, length_scale=length_scale)
    intro_part1_audio = AudioSegment.from_file(intro_part1_path)
    os.remove(intro_part1_path)

    chime_mid_intro_audio = AudioSegment.silent(duration=500, frame_rate=sample_rate)
    if sound_mode in ["With Effects", "Only Effects"] and "chime" in current_effects_list and "chime" in sound_mixer.sound_effects:
        chime_effect = sound_mixer.sound_effects["chime"]
        if chime_effect.frame_rate != sample_rate:
            chime_effect = chime_effect.set_frame_rate(sample_rate)
        chime_mid_intro_audio = chime_effect - 15
        current_effects_list.remove("chime")

    print("🔊 Generating introductory speech part 2...")
    intro_part2_path = generate_tts_wav(intro_part2_text, length_scale=length_scale)
    intro_part2_audio = AudioSegment.from_file(intro_part2_path)
    os.remove(intro_part2_path)

    full_intro_narration_audio = intro_part1_audio + chime_mid_intro_audio + intro_part2_audio
    final_intro_audio = full_intro_narration_audio

    if sound_mode in ["With Effects", "Only Effects"] and "soft_background_music" in current_effects_list and "soft_background_music" in sound_mixer.sound_effects:
        bgm_effect = sound_mixer.sound_effects["soft_background_music"]
        if bgm_effect.frame_rate != sample_rate:
            bgm_effect = bgm_effect.set_frame_rate(sample_rate)
        while len(bgm_effect) < len(full_intro_narration_audio):
            bgm_effect += bgm_effect
        bgm_effect = (bgm_effect[:len(full_intro_narration_audio)]) - 20
        final_intro_audio = final_intro_audio.overlay(bgm_effect)
        current_effects_list.remove("soft_background_music")

    print("\n🔊 Generating main lesson speech segment by segment...")
    main_lesson_audio_segments = []
    used_segment_effects = set()

    for i, segment in enumerate(lesson_segments):
        print(f"  Generating audio for segment {i + 1}/{len(lesson_segments)}: '{segment['sentence'][:50]}...'")
        sentence_audio_path = generate_tts_wav(segment["sentence"], length_scale=length_scale)
        sentence_audio_segment = AudioSegment.from_file(sentence_audio_path)
        os.remove(sentence_audio_path)

        processed_segment_audio = sentence_audio_segment

        if sound_mode == "With Effects" and segment["associated_sound_effects"]:
            print(f"    Overlaying specific sound effects: {segment['associated_sound_effects']}")
            for effect_name in segment["associated_sound_effects"]:
                if effect_name in sound_mixer.sound_effects:
                    effect = sound_mixer.sound_effects[effect_name]
                    if effect.frame_rate != sample_rate:
                        effect = effect.set_frame_rate(sample_rate)
                    effect = effect - 15
                    if len(effect) > len(processed_segment_audio):
                        effect = effect[:len(processed_segment_audio)]
                    processed_segment_audio = processed_segment_audio.overlay(effect, position=0)
                    used_segment_effects.add(effect_name)

        main_lesson_audio_segments.append(processed_segment_audio)

    # Remove segment-specific effects from general effect list to avoid duplicate playback later.
    current_effects_list = [eff for eff in current_effects_list if eff not in used_segment_effects]

    main_lesson_combined_audio = AudioSegment.silent(duration=0, frame_rate=sample_rate)
    if main_lesson_audio_segments:
        main_lesson_combined_audio = sum(main_lesson_audio_segments)

    combined_narration_audio = final_intro_audio + main_lesson_combined_audio

    print("\n🔊 Generating closing speech...")
    closing_path = generate_tts_wav("That's all for this section. Thank you for using me.", length_scale=length_scale)
    closing_audio = AudioSegment.from_file(closing_path)
    os.remove(closing_path)
    combined_narration_audio += closing_audio

    with tempfile.NamedTemporaryFile(delete=False, suffix=".wav") as f:
        base_audio_path = f.name
        combined_narration_audio.export(base_audio_path, format="wav")

    final_audio_to_play = base_audio_path

    if sound_mode == "With Effects":
        print("🎵 Mixing narration with remaining general sound effects...")
        mixed_audio = AudioSegment.from_file(base_audio_path)
        general_effects_to_add = [eff for eff in current_effects_list if eff in sound_mixer.sound_effects]

        if general_effects_to_add:
            interval_ms = len(mixed_audio) / (len(general_effects_to_add) + 1)
            current_position_ms = interval_ms

            for effect_name in general_effects_to_add:
                effect = sound_mixer.sound_effects[effect_name]
                if effect.frame_rate != sample_rate:
                    effect = effect.set_frame_rate(sample_rate)
                effect = effect - 20
                if current_position_ms + len(effect) < len(mixed_audio):
                    mixed_audio = mixed_audio.overlay(effect, position=current_position_ms)
                current_position_ms += interval_ms

        with tempfile.NamedTemporaryFile(delete=False, suffix=".wav") as f:
            final_audio_to_play = f.name
            mixed_audio.export(final_audio_to_play, format="wav")

    elif sound_mode == "Only Effects":
        print("🎵 Generating only sound effects...")
        only_effects_audio = AudioSegment.silent(duration=len(combined_narration_audio), frame_rate=sample_rate)

        all_possible_effects = set(lesson_data.get("sound_effects_list", []))
        for segment in lesson_segments:
            all_possible_effects.update(segment["associated_sound_effects"])

        if "chime" in lesson_data.get("sound_effects_list", []) and "chime" in sound_mixer.sound_effects:
            chime_effect = sound_mixer.sound_effects["chime"]
            if chime_effect.frame_rate != sample_rate:
                chime_effect = chime_effect.set_frame_rate(sample_rate)
            only_effects_audio = only_effects_audio.overlay(chime_effect - 15, position=len(intro_part1_audio))

        if "soft_background_music" in lesson_data.get("sound_effects_list", []) and "soft_background_music" in sound_mixer.sound_effects:
            bgm_effect = sound_mixer.sound_effects["soft_background_music"]
            if bgm_effect.frame_rate != sample_rate:
                bgm_effect = bgm_effect.set_frame_rate(sample_rate)
            while len(bgm_effect) < len(only_effects_audio):
                bgm_effect += bgm_effect
            bgm_effect = bgm_effect[:len(only_effects_audio)] - 20
            only_effects_audio = only_effects_audio.overlay(bgm_effect)

        effects_for_only_mode = [
            eff for eff in all_possible_effects
            if eff in sound_mixer.sound_effects and eff not in ["chime", "soft_background_music"]
        ]

        if effects_for_only_mode:
            interval_ms = len(only_effects_audio) / (len(effects_for_only_mode) + 1)
            current_position_ms = interval_ms
            for effect_name in effects_for_only_mode:
                effect = sound_mixer.sound_effects[effect_name]
                if effect.frame_rate != sample_rate:
                    effect = effect.set_frame_rate(sample_rate)
                effect = effect - 15
                if current_position_ms + len(effect) < len(only_effects_audio):
                    only_effects_audio = only_effects_audio.overlay(effect, position=current_position_ms)
                current_position_ms += interval_ms

        with tempfile.NamedTemporaryFile(delete=False, suffix=".wav") as f:
            final_audio_to_play = f.name
            only_effects_audio.export(final_audio_to_play, format="wav")

    else:
        print("🔊 Playing narration only...")

    display(Audio(final_audio_to_play, autoplay=True))

    if os.path.exists(base_audio_path):
        os.remove(base_audio_path)
    if final_audio_to_play != base_audio_path and os.path.exists(final_audio_to_play):
        os.remove(final_audio_to_play)

    print("\n📖 Lesson Text Preview:")
    print(text_to_speak[:300] + "...")
```
